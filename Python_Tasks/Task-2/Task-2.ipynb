## Q1. Find the Shortest Word in a Sentence


```python
# Observe:input(scentence)-->output(one word)
## for example input is "Python makes coding easy" excepted output is "easy".
# understand: The problem is asking me to find the shortewst word in the scentence.
#Break:scentence-->split the scentence-->loop over the words-->calculate the length-->find shortest word.
#logic:input (scentence)-->split the scentence-->shortest_word =first_word-->for loop over the words-->update the shortest word if len(shortest_word)>i
#code:
scentence=eval(input("give the scentence"))
a=scentence.split()
shortest_word=a[0]
for i in range(len(a)):
    if len(a[i])<len(shortest_word):
        shortest_word=a[i]
print(shortest_word)        
#Test: executed the code and observe that expected output is same as original output      
```

## Q6. Count Vowels in a String


```python
#observe:input is a string and output is a integer
#understand:The problem asking to count the vowels in a string
#for example string="Education",vowels are [a,e,i,o,u] and we have to count the vowels in the string for this input expected output is 5.
#Break:input(string)-->convert into lowercase-->loop-->count
#logic:input(string)-->.lower-->count=0-->loop over string--checking vowels in string using if and logic OR-->if vowel is present incerease count+=1-->get count
#code:
string=eval(input("Give the string"))
string=string.lower()
count=0
for i in string:
    if i=="a" or i=="e" or i=="i" or i=="o" or i=="u":
        count+=1
print(count)    
#Test: executed the code and observe that expected output is same as original output 
```

## Q7. Count Consonants in a String

```python
#observe:input is a string and output is a integer
#understand:The problem asking to count the consonants in a string
#for example string="Python 123!",vowels are [a,e,i,o,u] and we have to count the consonants in the string for this input expected output is 5.
#Break:input(string)-->convert into lowercase-->loop-->count
#logic:input(string)-->.lower-->count=0-->loop over string--checking consonants in string using if and logic NOT-->if vowel is not present increase count+=1-->get count
#code:
string=eval(input("Give the string"))
string=string.lower()
vowels=["a","e","i","o","u"]
count=0
for i in string:
    if i not in vowels and i.isalpha()==True:
        count+=1
    
print(count)    
#Test: executed the code and observe that expected output is same as original output 
```

```python
#observe:input is a string and output is a string
#understand:The problem asking to find the most frequent character in a string.
#for example string="banana",and output="a" we have count each character in a string and find the most frequent character.
#Break:input(string)-->convert into lowercase-->loop-->count the occurance
#logic:input(string)-->.lower--using count built-in count the occurance of character.
#code:
string=eval(input("give the word"))
string=string.lower()
l1=[]
l2=[]
for i in string:
    if i not in l1:
        l1.append(i)
    else:
        l2.append(i)
max_count=0
letter=[]
for i in l2:
    if l2.count(i)>max_count:
        max_count=l2.count(i)
        letter.append(i)
print(letter)
#Test: executed the code and observe that expected output is same as original output    
    
    
```
