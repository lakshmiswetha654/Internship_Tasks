Q1. Find the Frequency of Each Word in a Sentence

```python
#observe:input=scentence output=dict understand:asking to get words with frequency in dictionary form. break:using function 
#first convert into lowercase and split the scentence by using two list and with help of loop and not in operator and using append 
#unique values are append into one list and duplicate values into other list by using the unique words count the frequency of words in the scentence.
#input="python is easy and python is powerful"
#expected_output={'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'powerful': 1}
def frequency_count(a):
    a=a.lower()
    a=a.split()
    
    dict={}
    l1=[]
    l2=[]
    for i in a:
        if i not in l1:
            l1.append(i)
        else:
            l2.append(i)

    for i in l1:
        dict[i]=a.count(i)
    return dict 
print("Q1:", frequency_count("python is easy and python is powerful"))
print("Q1 additional:",frequency_count("hello hello world"))   
```

Q2. Find the Frequency of Each Character in a Word

```python
#observe:input=word output=dict understand:asking to get character with frequency in dictionary form.using function 
#first convert into lowercase with two empty list and with help of loop and not in operator and using append 
#unique characters are append into one list and duplicate characters into other list by using the unique character count the frequency of characters in the word.
#input="banana"
#expected_output={'b': 1, 'a': 3, 'n': 2}
def character_frequency(a):
    a=a.lower()
    l1=[]
    l2=[]
    dict={}
    for i in range(len(a)):
        if a[i] not in l1:
            l1.append(a[i])
        else:
            l2.append(a[i])
    for i in l1:
        dict[i]=a.count(i) 
    return dict
print("Q2:", character_frequency("banana"))
print("Q2 additional:", character_frequency("apple"))
```

 ### Q3. Find the First Non-Repeating Character in a Word

```python
#observe:input=word output=letter understand:asking to get non repeating character from the word.using function 
#first convert into lowercase with help of two empty list and with help of loop and not in operator and using append 
#unique characters are append into one list and duplicate characters into other list by using the unique character count the frequency of characters in the word
# using condition if count=1 return the letter.
#input="swiss"
#output='w'
def non_repeating_character(a):
    a=a.lower()
    l1=[]
    l2=[]
    for i in range(len(a)):
        if a[i] not in l1:
            l1.append(a[i])
        else:
            l2.append(a[i])
    for i in l1:
        if a.count(i)==1:
            return i
            break
print("Q3:",non_repeating_character("swiss"))
print("Q3 additional:",non_repeating_character("aabb"))               
```

### Q4. Check Whether Two Strings Are Anagrams

```python
#observe:input=two words output= bool(True/False understand:asking to find wheater the given two words are anagram are not.using function 
#first convert the two words into lowercase and check both sorted words are same are not. if both the sorted words are same then it return True or False if sorted words are not same.
#input="listen", "silent"
#output=True
def is_anagram(a,b):
    a=a.lower()
    b=b.lower()
    if sorted(a)==sorted(b):
        return("True")
    else:
        return("False")
print("Q4:", is_anagram("listen", "silent"))
print("Q4 additional:", is_anagram("hello", "world"))   
```

### Q5. Find the First Repeating Character in a String

```python
#input=string output=letter ,problem asking to find the first repeating character in a string. using function first 
# convert the word into lowercase and by using two empty list with help of not in operator and using append builtin 
#unique character is append into one list and duplicates into other list and from the duplicate list return the first letter using indexing.
#input="programming" output='r'
def first_repeating_character(a):
    a=a.lower()
    l1=[]
    l2=[]
    for i in range(len(a)):
        if a[i] not in l1:
            l1.append(a[i])
        else:
            l2.append(a[i])
    if len(l2)==0:
        return "None"
    else:
        return l2[0]
print("Q5:", first_repeating_character("programming"))
print("Q5 additional:", first_repeating_character("abcde"))
```

### Q6. Move All Zeros to the End of a List

```python
#input=list and output=list .the problem asking to move all zeros to the enf of the list.
#using function and with the help of two empty lists, loop, conditional statement and append builtin 
#if i=0 then append the zero into one list and other values into other list 
#at last using extend builtin join both values list with zero list by this we can move the all zeros to the end of a list.
#input=[0, 3, 0, 1, 2, 0, 4] output=[3,1,2,0,0,0]
def zeros_shifting(a):
    num=[]
    z=[]
    for i in a:
        if i==0:
            z.append(i)
        else:
            num.append(i)
    num.extend(z)
    return num
print("Q6:",zeros_shifting([0, 3, 0, 1, 2, 0, 4]))
print("Q6 additional:",zeros_shifting([1, 0, 2]))
```

### Q7. Find the Second Highest Distinct Number in a List

```python
#input=list and output =number . the problem asking to find the second highest distinct number in a list. with the help of functions 
#empty list, using not in operator and append builtin and alo using conditional statement, sort biltin, loop.
#using not in operator if number not in empty list then append the number into the list those are unique number and sorting that list and using 
#using indexing for returning second highest number from a list.
#input=[10, 20, 40, 30, 40, 20]
#output=30
def second_highest(nums):
    unique_numbers = []
    for number in nums:
        if number not in unique_numbers:
            unique_numbers.append(number)
    unique_numbers.sort(reverse=True)
    if len(unique_numbers) < 2:
        return None
    return unique_numbers[1]
print("Q7:", second_highest([10, 20, 40, 30, 40, 20]))
print("Q7 additional:", second_highest([5, 5, 5]))
```

### Q8. Remove Duplicates Without Changing the Original Order

```python
#input=list output=list. the problem asking to remove duplicates without changing the original order
#using function ,two empty lists not in operator, append builtin,loop
# using for loop if number not in the list then append the number into the l1 list and if it presnt then append into other list
# to get list without duplicates then return first list.
# input=[4, 2, 4, 5, 2, 7, 5] 
#output=[4, 2, 5, 7]
def remove_duplicates(a):
    l1=[]
    l2=[]
    for i in a:
        if i not in l1:
            l1.append(i)
        else:
            l2.append(i)
    return l1
print("Q8:", remove_duplicates([4, 2, 4, 5, 2, 7, 5]))
print("Q8 additional:", remove_duplicates([1, 1, 2, 3, 2]))                     
```

### Q9 Find the Missing Number from 1 to N

```python
#input=list output=number. the problem asking to find the missing number from 1 to N
#using loop over the range of last number in the list and by using not in operator we can return missing value.
#input=[1, 2, 4, 5, 6]
#output=3
def missing_value(a):
    for i in range(1,a[-1]+1):
        if i not in a:
            return i
print("Q9:", missing_value([1, 2, 4, 5, 6]))
print("Q9 additional:", missing_value([1, 3, 4]))
        
```

#### 10. Find Two Numbers Whose Sum Equals a Target

```python
#input=list output=list. the problem asking to find the two numbers whose sum equal to target.
#by using function and taking target value, with using outer and inner loop at range of len(a) and using condition to find the two numbers 
#whose sum equal to target using extend builtin we can get the list of those numbers.
#input=[2, 7, 11, 15]
#output=[2,7]
def add(a):
    target=9
    num=[]
    for i in a:
        for j in range(1,len(a)):
            if a[j]+i==target:
                num.extend([i,a[j]])
    return num
print("Q10:", add([2, 7, 11, 15]))
print("Q10 additional:", add([1, 2, 3]))
```

### Q12. Find Common Elements Between Two Lists

```python
#input=list,output=list. the problem using to find the common elements between two list.
#using loops, append builtin, conditional statements.
# loop over the one list if num of a in b then append num into the list and return the list.
#input=[1, 2, 3, 4, 5], [3, 4, 5, 6, 7]
#output=[3, 4, 5]

def common_element(a,b):
    common=[]
    for i in a:
        if i in b:
            common.append(i)
    return common
print("Q12:", common_element([1, 2, 3, 4, 5], [3, 4, 5, 6, 7]))
print("Q12 additional:", common_element([1, 1, 2], [2, 1]))
```

### Q13. Rotate a List to the Right by K Positions

```python
#input=list output=list. the problem is asking to rotate the list to the right by K position.
#using function and taking position(K) at where the list to be reversed and taking first list upto the position mention 
#and second list from the mention position. using arthematic operator anding both first and second list but second list to be added in reverse.
#input=[1, 2, 3, 4, 5] ,position(K)=2
#output=[1,2,5,4,3]
def reverse(a):
    K=2
    first=a[:K]
    next=a[K:]
    return first+next[::-1]
print("Q13:",reverse([1, 2, 3, 4, 5]))
print("Q13 additional:", reverse([1, 2, 3]))               
```

### Q14. Find the Maximum Sum of Two Consecutive Elements

```python
#input=list output=number. the problem asking to find maximum sum of two consecutives elements.
#using functions, loop over the range of len(list)-1, conditional statement.
#first assume the max_sum=0 and check the numbers in the list whose sum is maximum.
#input=[4, 7, 2, 9, 5]
#output=14
def maximum_sum(a):
    max_sum=0
    for i in range(len(a)-1):
        if a[i]+a[i+1]>max_sum:
            max_sum=a[i]+a[i+1]
    return( max_sum )
print("Q14:", maximum_sum([4, 7, 2, 9, 5]))
print("Q14 additional:", maximum_sum([-5, -2, -7]))
 
```

### Q15. Find All Pairs of Numbers Whose Sum Equals a Target

```python
#input=list output=list. the problem asking to find all pairs of numbers whose sum equal to target.
#using functions,loop,append builtin,sorted.
#for loop over the range of len(a). if whose two numbers equal to target then append the numbers into the empty list.
# use another loop to remove the duplicates from the list and sort the list to get numbers in ascendding order.
#input=[2, 4, 3, 5, 7, 8, 1]
#output=[[1, 8], [2, 7], [4, 5]]
def add(a):
    target=9
    num=[]
    nums=[]
    for i in a:
        for j in range(len(a)):
            if a[j]+i==target:
                num.append([i,a[j]])
    for i in range (len(num)):
        if num[i][::] not in nums and num[i][::-1] not in nums:
            nums.append(sorted(num[i]))
    
    return sorted(nums) 
print("Q15:", add([2, 4, 3, 5, 7, 8, 1]))
print("Q15 additional:", add([1, 2, 3]))
       
```
